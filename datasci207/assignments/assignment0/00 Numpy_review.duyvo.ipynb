#### <span style="color:chocolate">Exercise 8:</span> Joining arrays (10 points)

Provide 8 examples that use the methods <span style="color:cornflowerBlue">np.concatenate()</span> and <span style="color:cornflowerBlue">np.stack()</span>. Note: create 4 examples using each method.

```python
input_1x2_a_array = np.array([1, 2])
input_1x2_b_array = np.array([3, 4])
input_1x2_c_array = np.array([5, 6])
input_2x2_a_array = np.array([[1, 2], [3, 4]])
input_2x2_b_array = np.array([[5, 6], [7, 8]])

# Concatenate 2 1D arrays
c1_array = np.concatenate((input_1x2_a_array, input_1x2_b_array))
print("c1_array:\n", c1_array)

# Concatenate vertically (2D array)
c2_array = np.concatenate((input_2x2_a_array, input_2x2_b_array), axis=0)
print("c2_array:\n", c2_array)

# Concatenate horizontally (2D array)
c3_array = np.concatenate((input_2x2_a_array, input_2x2_b_array), axis=1)
print("c3_array:\n", c3_array)

# Concatenate 3 1D arrays
c4_array = np.concatenate((input_1x2_a_array, input_1x2_b_array, input_1x2_c_array))
print("c4_array:\n", c4_array)

# Create stacks as with layers
s1_stack = np.stack((input_2x2_a_array, input_2x2_b_array), axis=0)
print("s1_stack:\n", s1_stack)

# Create stacks with alternating rows
s2_stack = np.stack((input_2x2_a_array, input_2x2_b_array), axis=1)
print("s2_stack:\n", s2_stack)

# Create stacks with alternating columns
s3_stack = np.stack((input_2x2_a_array, input_2x2_b_array), axis=2)
print("s3_stack:\n", s3_stack)

# Create 2D arrays from 2 1D arrays
s4_stack = np.stack((input_1x2_a_array, input_1x2_b_array), axis=0)
print("s4_stack:\n", s4_stack)
```

#### <span style="color:chocolate">Numpy broadcasting</span> 

The term broadcasting describes how NumPy treats arrays during arithmetic operations. You can find more here:  https://numpy.org/doc/stable/user/basics.broadcasting.html

```python
# multiplying an array with a scalar
years = np.arange(1991, 2018, 1)
years * 1
```

```python
# adding a scalar
years + 2
```

```python
# many arithmetic operations are applied element wise on a pair of arrays
a = np.arange(3)
print('Array a:', a)
b = np.arange(3)
print('Array b:', b)
print('Sum of a and b:', a+b)
```

```python
# another example
a = np.ones((3, 3))
print('Array a:', a)
b = np.arange(3)
print('\nArray b:', b)
print('\nSum of a and b:', a+b)
```

#### <span style="color:chocolate">Exercise 9:</span> Numpy array standardization (10 points)

Consider the following (10, 4) array:

```python
my_array = np.random.random((10, 4))
print('my_array shape: ', my_array.shape)
my_array
```

Peform the following operations on this array:
1. Compute and print the mean and std for each column (note: set keepdims=True for proper broadcasting).
2. Standardize and print the array and its shape using the mean and std you computed above.

```python
my_array_mean = my_array.mean(axis=0, keepdims=True)
my_array_std = my_array.std(axis=0, keepdims=True)
print("Column means of my_array: ", my_array_mean)
print("Column std of my_array: ", my_array_std)

standardized_array = (my_array - my_array_mean) / my_array_std
print("standardized_array:\n", standardized_array)
print("Shape of standardized_array: ", standardized_array.shape)
```
