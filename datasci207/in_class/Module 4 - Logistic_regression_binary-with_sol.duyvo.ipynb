```python
def confusion_matrix_plot(conf_matrix):
    """ Confusion matrix matplotlib plot
    # param conf_matrix: nested list of TP, TN, FP, FN
    # return: None
    """
    fig, ax = plt.subplots(figsize=(2.5, 2.5))
    ax.matshow(conf_matrix, cmap=plt.cm.Blues, alpha=0.3)
    for i in range(conf_matrix.shape[0]):
        for j in range(conf_matrix.shape[1]):
            ax.text(x=j, y=i, s=conf_matrix[i, j], va='center', ha='center')

    plt.xlabel('Predicted label')
    plt.ylabel('True label')

    plt.tight_layout()
    plt.show()
```

```python
def draw_neural_net(ax, layer_sizes, coefs_, intercepts_,
                    left=0.1, right=.5, bottom=0.1, top=.5):
    '''
    Draw a neural network cartoon using matplotilb.
    Adapted: https://gist.github.com/craffel/2d727968c3aaebd10359
    
    Args:
        - ax : matplotlib.axes.AxesSubplot
            The axes on which to plot the cartoon (get e.g. by plt.gca())
        - layer_sizes : list of int
            List of layer sizes, including input and output dimensionality
    '''
    n_layers = len(layer_sizes)
    v_spacing = (top - bottom)/float(max(layer_sizes))
    h_spacing = (right - left)/float(len(layer_sizes) - 1)
    
    # Input-Arrows
    layer_top_0 = v_spacing*(layer_sizes[0] - 1)/2. + (top + bottom)/2.
    for m in range(layer_sizes[0]):
        plt.arrow(left-0.18, layer_top_0 - m*v_spacing, 0.12, 0,  lw=1, head_width=0.01, head_length=0.02)
    
    # Nodes
    for n, layer_size in enumerate(layer_sizes):
        layer_top = v_spacing*(layer_size - 1)/2. + (top + bottom)/2.
        for m in range(layer_size):
            circle = plt.Circle((n*h_spacing + left, layer_top - m*v_spacing), v_spacing/8.,
                                color='w', ec='k', zorder=4)
            if n == 0:
                plt.text(left-0.125, layer_top - m*v_spacing, r'$X_{'+str(m+1)+'}$', fontsize=15)
            elif (n_layers == 3) & (n == 1):
                plt.text(n*h_spacing + left+0.00, layer_top - m*v_spacing+ (v_spacing/8.+0.01*v_spacing), r'$H_{'+str(m+1)+'}$', fontsize=15)
            elif n == n_layers -1:
                plt.text(n*h_spacing + left+0.10, layer_top - m*v_spacing, r'$y_{'+str(m+1)+'}$', fontsize=15)
            ax.add_artist(circle)
    # Bias-Nodes
    for n, layer_size in enumerate(layer_sizes):
        if n < n_layers -1:
            x_bias = (n+0.5)*h_spacing + left
            y_bias = top + 0.005
            circle = plt.Circle((x_bias, y_bias), v_spacing/8., color='w', ec='k', zorder=4)
            plt.text(x_bias-(v_spacing/8.+0.10*v_spacing-0.01), y_bias, r'$1$', fontsize=15)
            ax.add_artist(circle)   
    # Edges
    # Edges between nodes
    for n, (layer_size_a, layer_size_b) in enumerate(zip(layer_sizes[:-1], layer_sizes[1:])):
        layer_top_a = v_spacing*(layer_size_a - 1)/2. + (top + bottom)/2.
        layer_top_b = v_spacing*(layer_size_b - 1)/2. + (top + bottom)/2.
        for m in range(layer_size_a):
            for o in range(layer_size_b):
                line = plt.Line2D([n*h_spacing + left, (n + 1)*h_spacing + left],
                                  [layer_top_a - m*v_spacing, layer_top_b - o*v_spacing], c='k')
                ax.add_artist(line)
                xm = (n*h_spacing + left)
                xo = ((n + 1)*h_spacing + left)
                ym = (layer_top_a - m*v_spacing)
                yo = (layer_top_b - o*v_spacing)
                rot_mo_rad = np.arctan((yo-ym)/(xo-xm))
                rot_mo_deg = rot_mo_rad*180./np.pi
                xm1 = xm + (v_spacing/8.+0.05)*np.cos(rot_mo_rad)
                if n == 0:
                    if yo > ym:
                        ym1 = ym + (v_spacing/8.+0.08)*np.sin(rot_mo_rad)
                    else:
                        ym1 = ym + (v_spacing/8.+0.1)*np.sin(rot_mo_rad) + .01
                else:
                    if yo > ym:
                        ym1 = ym + (v_spacing/8.+0.08)*np.sin(rot_mo_rad)
                    else:
                        ym1 = ym + (v_spacing/8.+0.08)*np.sin(rot_mo_rad)
                plt.text(xm1, ym1,
                         str(round(coefs_[n][m, o],2)),
                         rotation = rot_mo_deg,
                         fontsize = 10)
    # Edges between bias and nodes
    for n, (layer_size_a, layer_size_b) in enumerate(zip(layer_sizes[:-1], layer_sizes[1:])):
        if n < n_layers-1:
            layer_top_a = v_spacing*(layer_size_a - 1)/2. + (top + bottom)/2.
            layer_top_b = v_spacing*(layer_size_b - 1)/2. + (top + bottom)/2.
        x_bias = (n+0.5)*h_spacing + left
        y_bias = top + 0.005 
        for o in range(layer_size_b):
            line = plt.Line2D([x_bias, (n + 1)*h_spacing + left],
                          [y_bias, layer_top_b - o*v_spacing], c='k')
            ax.add_artist(line)
            xo = ((n + 1)*h_spacing + left)
            yo = (layer_top_b - o*v_spacing)
            rot_bo_rad = np.arctan((yo-y_bias)/(xo-x_bias))
            rot_bo_deg = rot_bo_rad*180./np.pi
            xo2 = xo - (v_spacing/8.+0.01)*np.cos(rot_bo_rad)
            yo2 = yo - (v_spacing/8.+0.01)*np.sin(rot_bo_rad)
            xo1 = xo2 -0.08 * np.cos(rot_bo_rad)
            yo1 = yo2 -0.05 * np.sin(rot_bo_rad)
            plt.text(xo1, yo1,
                 str(round(intercepts_[n][o],2)),
                 rotation = rot_bo_deg,
                 fontsize = 10)    
                
    # Output-Arrows
    layer_top_0 = v_spacing*(layer_sizes[-1] - 1)/2. + (top + bottom)/2.
    for m in range(layer_sizes[-1]):
        plt.arrow(right+0.015, layer_top_0 - m*v_spacing, 0.16*h_spacing, 0,  lw =1, head_width=0.01, head_length=0.02)
```

### Step 6: Analysis - Logistic Regression with two features
---

``from scratch``

We will start by fitting the training data using <span style="color:chocolate"> gradient descent</span> and we will implement this algorithm from scratch. We will then use TensorFlow to see how the predictions compare. <br>

Below you can see a visual representation of the difference between the logistic regression and the linear regression model.

```python
Image(filename='./images/linear_logistic.png', width=500)
```

As you can see, comparing to the Linear Regression model, we need to <span style="color:chocolate"> redefine the shape of the activation function and the threshold function</span> to fit a logistic regression model.

Let j represent features, j = [1,...,m]

Let i represent observations, i = [1, ..., n]


``Net input function``:

z= $w_{0} \cdot 1 + w_{2} \cdot x_{1} + w_{2} \cdot x_{2} + ... + w_{m} \cdot x_{m}$

``Sigmoid activation function``:

$\phi(z)= \frac{1}{1+exp(-z)}$


`` Compute error``:

$(y\_actual_{i} - \phi(z_{i}))$, , where $\phi(z_{i})$ is $y\_pred_{i}$

``Update weights``:

$w_{j} := w_{j} + \Delta w_{j}, \forall j \in {1,...,m}$

$\Delta w_{j} = \eta \sum _{i}(y\_actual_{i} - \phi(z_{i}))\cdot x_{i,j}, \forall j \in {1,...,m}, \forall i \in {1,...,n}$


``Threshold function``: (changed to reflect a logistic outcome)

$\tau(z)  = 
\begin{cases}
1,  if z \geq 0.5,\\
0, otherwise
\end{cases}
$



`` Note on Gradient Descent``:

The weights are updated using **all training examples** (see the summation operator in the equation $\Delta w_{j}$)

Now let's code the math:

```python
# Step 1: Initialize hyperparameters
eta = 0.1
n_iter = 20
random_state = 1

def fit(X, y):
    # Step 2: Initialize weights (small random numbers)
    rgen = np.random.RandomState(random_state)
    w_ = rgen.normal(loc=0.0, scale=0.01, size=1 + X.shape[1])

    # Step 3: Initialize cost list
    cost_ = []
    
    # Step 4: Compute net input, sigmoid activation, update weights, track cost
    for iter_ in range(n_iter):
        # Compute the net input function
        net_input = np.dot(X, w_[1:]) + w_[0]

        # Compute the sigmoid activation function
        temp_y_pred = 1. / (1. + np.exp(-np.clip(net_input, -250, 250)))
        
        # Compute error
        errors = (y - temp_y_pred)

        # Update weights using all training examples (batch gradient descent)
        w_[1:] += eta * X.T.dot(errors)
        w_[0] += eta * errors.sum()

        # Compute the log-likelihood cost function
        epsilon = 1e-15  # To prevent log(0)
        temp_y_pred = np.clip(temp_y_pred, epsilon, 1 - epsilon)
        cost = (-y.dot(np.log(temp_y_pred)) - ((1 - y).dot(np.log(1 - temp_y_pred)))) / X.shape[0]
        cost_.append(cost)
        
    return cost_, w_

# Step 5: Predict y using threshold function
def predict(X, w_):
    # Compute net input
    net_input = np.dot(X, w_[1:]) + w_[0]
    # Apply sigmoid function
    y_pred = 1. / (1. + np.exp(-np.clip(net_input, -250, 250)))
    # Apply threshold to produce binary predictions
    y_pred = np.where(y_pred >= 0.5, 1, 0)
    return y_pred
```

Model fit

```python
cost_train, w_train = fit(X_train_std, y_train)
cost_val, w_val = fit(X_val_std, y_val)

# plot cost
plt.plot((range(1, len(cost_train) + 1)), cost_train, marker='o', label='train data')
plt.plot((range(1, len(cost_val) + 1)), cost_val, marker='o', label='val data');

plt.xlabel('epochs (n_iter)');
plt.ylabel('Logistic cost'); # number of times y_actual != y_pred
plt.legend()

plt.title('Logistic convergence with BGD. Learning rate = ' + str(eta));
```

```python
# SOLUTION: Plot cost for validation data using training weights
# and analyze convergence with different hyperparameters

# First, let's compute validation cost using the training weights
def compute_validation_cost(X_val, y_val, w_):
    """Compute the cost for validation data using given weights"""
    # Compute net input
    net_input = np.dot(X_val, w_[1:]) + w_[0]
    # Apply sigmoid
    y_pred = 1. / (1. + np.exp(-np.clip(net_input, -250, 250)))
    # Compute log-likelihood cost
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    cost = (-y_val.dot(np.log(y_pred)) - ((1 - y_val).dot(np.log(1 - y_pred)))) / X_val.shape[0]
    return cost

# Create a version that tracks weights at each iteration
def fit_track_weights(X, y, eta=0.1, n_iter=20, random_state=1):
    """Modified fit function that returns weights at each iteration"""
    rgen = np.random.RandomState(random_state)
    w_ = rgen.normal(loc=0.0, scale=0.01, size=1 + X.shape[1])
    
    cost_ = []
    weights_ = [w_.copy()]  # Track weights at each iteration
    
    for iter_ in range(n_iter):
        net_input = np.dot(X, w_[1:]) + w_[0]
        temp_y_pred = 1. / (1. + np.exp(-np.clip(net_input, -250, 250)))
        errors = (y - temp_y_pred)
        
        w_[1:] += eta * X.T.dot(errors)
        w_[0] += eta * errors.sum()
        weights_.append(w_.copy())
        
        epsilon = 1e-15
        temp_y_pred = np.clip(temp_y_pred, epsilon, 1 - epsilon)
        cost = (-y.dot(np.log(temp_y_pred)) - ((1 - y).dot(np.log(1 - temp_y_pred)))) / X.shape[0]
        cost_.append(cost)
    
    return cost_, w_, weights_

# Test 1: Original hyperparameters
print("\nTest 1: Original hyperparameters (eta=0.1, n_iter=20)")
cost_train_1, w_train_1, weights_1 = fit_track_weights(X_train_std, y_train, eta=0.1, n_iter=20)

# Compute validation costs using the training weights from each iteration
cost_val_1 = []
for w_ in weights_1[:-1]:  # Skip the initial weights
    cost_val_1.append(compute_validation_cost(X_val_std, y_val, w_))

# Plot training vs validation cost
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Plot 1: Original hyperparameters
ax = axes[0]
ax.plot(range(1, len(cost_train_1) + 1), cost_train_1, marker='o', label='Training cost', linewidth=2)
ax.plot(range(1, len(cost_val_1) + 1), cost_val_1, marker='s', label='Validation cost', linewidth=2)
ax.set_xlabel('Epochs (n_iter)')
ax.set_ylabel('Logistic cost')
ax.set_title('Original: eta=0.1, n_iter=20')
ax.legend()
ax.grid(True, alpha=0.3)

# Test 2: More iterations (n_iter=100)
print("\nTest 2: More iterations (eta=0.1, n_iter=100)")
cost_train_2, w_train_2, weights_2 = fit_track_weights(X_train_std, y_train, eta=0.1, n_iter=100)

cost_val_2 = []
for w_ in weights_2[:-1]:
    cost_val_2.append(compute_validation_cost(X_val_std, y_val, w_))

# Plot 2: More iterations
ax = axes[1]
ax.plot(range(1, len(cost_train_2) + 1), cost_train_2, marker='o', label='Training cost', linewidth=2)
ax.plot(range(1, len(cost_val_2) + 1), cost_val_2, marker='s', label='Validation cost', linewidth=2)
ax.set_xlabel('Epochs (n_iter)')
ax.set_ylabel('Logistic cost')
ax.set_title('More iterations: eta=0.1, n_iter=100')
ax.legend()
ax.grid(True, alpha=0.3)

# Test 3: Different learning rate (eta=0.01)
print("\nTest 3: Different learning rate (eta=0.01, n_iter=100)")
cost_train_3, w_train_3, weights_3 = fit_track_weights(X_train_std, y_train, eta=0.01, n_iter=100)

cost_val_3 = []
for w_ in weights_3[:-1]:
    cost_val_3.append(compute_validation_cost(X_val_std, y_val, w_))

# Plot 3: Different learning rate
ax = axes[2]
ax.plot(range(1, len(cost_train_3) + 1), cost_train_3, marker='o', label='Training cost', linewidth=2)
ax.plot(range(1, len(cost_val_3) + 1), cost_val_3, marker='s', label='Validation cost', linewidth=2)
ax.set_xlabel('Epochs (n_iter)')
ax.set_ylabel('Logistic cost')
ax.set_title('Lower learning rate: eta=0.01, n_iter=100')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Analysis and observations
print("\n" + "="*60)
print("ANALYSIS AND OBSERVATIONS")
print("="*60)

print(f"\nTest 1 (eta=0.1, n_iter=20):")
print(f"  Final training cost: {cost_train_1[-1]:.6f}")
print(f"  Final validation cost: {cost_val_1[-1]:.6f}")
print(f"  Training vs Validation gap: {abs(cost_train_1[-1] - cost_val_1[-1]):.6f}")
print(f"  Model appears to be: ", end="")
if cost_val_1[-1] > cost_train_1[-1] * 1.2:
    print("OVERFITTING (validation cost much higher)")
elif abs(cost_train_1[-1] - cost_val_1[-1]) < 0.01:
    print("WELL-GENERALIZING (costs are similar)")
else:
    print("REASONABLE (costs are relatively close)")

print(f"\nTest 2 (eta=0.1, n_iter=100):")
print(f"  Final training cost: {cost_train_2[-1]:.6f}")
print(f"  Final validation cost: {cost_val_2[-1]:.6f}")
print(f"  Convergence: {'Better' if cost_train_2[-1] < cost_train_1[-1] else 'Worse'} than Test 1")

print(f"\nTest 3 (eta=0.01, n_iter=100):")
print(f"  Final training cost: {cost_train_3[-1]:.6f}")
print(f"  Final validation cost: {cost_val_3[-1]:.6f}")
print(f"  Impact of lower learning rate: ", end="")
if cost_train_3[-1] > cost_train_2[-1]:
    print("Slower convergence (higher cost)")
else:
    print("Better convergence or similar")
```

1. CONVERGENCE: The model should show decreasing training and validation costs
   over iterations. Both should stabilize, indicating convergence.

2. OVERFITTING DETECTION: If validation cost significantly exceeds training cost
   and diverges, the model is overfitting. Use this gap to tune hyperparameters.

3. LEARNING RATE EFFECT:
   - High eta (e.g., 0.1): Faster convergence but might oscillate or diverge
   - Low eta (e.g., 0.01): Slower convergence but more stable training
   
4. NUMBER OF ITERATIONS:
   - Few iterations: Model may not converge fully (high cost)
   - Many iterations: Better convergence, but diminishing returns beyond
     a certain point; check for overfitting

5. OPTIMAL STRATEGY: Compare training vs validation curves to find the sweet spot
   where both decrease and neither overfits. This is the "best" hyperparameter set.

Define and compile model

```python
def build_model(num_features):
    """Return a simple logistic regression model using the Keras Sequential API."""
  # Clear session and set a random seed for consistent behavior.
    tf.keras.backend.clear_session()
    tf.random.set_seed(1234)
    

  # Use Keras Sequential API to build a logistic regression
    model = keras.Sequential()
    
    # create input layer
    model.add(tf.keras.Input(shape=(num_features,),
         name='Input'
    ))
   
    
    model.add(keras.layers.Dense(
        units=1,  
        use_bias=True,
        activation='sigmoid',
        #kernel_initializer='random_normal',
        #bias_initializer='zeros',
        kernel_initializer=initializers.RandomNormal(stddev=0.01),
        bias_initializer=initializers.RandomNormal(stddev=0.01),
        name="Output"
    ))
  
  # Use Binary Crossentropy as our loss and the SGD optimizer.
    model.compile(
        loss=tf.keras.losses.BinaryCrossentropy(),
        optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
        metrics=[tf.keras.metrics.BinaryAccuracy()]
    )
    return model

# Build a model
model = build_model(num_features=len(features))
model.summary()
```
